```python
import pandas as pd
import random
df = pd.read_csv("features.csv", low_memory=False)
reviews = []
labels = []
for _, row in df.iterrows():
    review_parts = []
    score = 0
    # Shipping Cost
    if row["shipping_cost"] <= 100:
        review_parts.append("Shipping cost was affordable.")
        score += 1
    elif row["shipping_cost"] >= 250:
        review_parts.append("Shipping charges were too expensive.")
        score -= 1
    # Distance
    if row["distance_km"] <= 100:
        review_parts.append("Delivery arrived quickly.")
        score += 1
    elif row["distance_km"] >= 300:
        review_parts.append("Delivery took longer than expected.")
        score -= 1
    # Express Delivery
    if bool(row["express_flag"]):
        review_parts.append("Express delivery was very convenient.")
        score += 1
    # Quantity
    if row["quantity"] >= 5:
        review_parts.append("Ordered several items together.")
        score += 0.5
    # Price
    if row["price"] >= 7000:
        review_parts.append("The product feels premium.")
        score += 0.5
    elif row["price"] <= 1000:
        review_parts.append("The product quality was disappointing.")
        score -= 1
    # Customer Spending
    if pd.notna(row.get("total_spend")):
        if row["total_spend"] > 50000:
            review_parts.append("I am a loyal customer and usually have a good experience.")
            score += 1
    # Purchase Frequency
    if pd.notna(row.get("total_orders")):
        if row["total_orders"] > 20:
            review_parts.append("I frequently order from this store.")
            score += 0.5
    # Default review if nothing added
    if len(review_parts) == 0:
        review_parts.append("The shopping experience was average.")
    review = " ".join(review_parts)
    # Determine sentiment
    if score >= 1:
        label = "positive"
    elif score <= -1:
        label = "negative"
    else:
        label = random.choice(["positive", "negative"])
    reviews.append(review)
    labels.append(label)
review_df = pd.DataFrame({
    "review": reviews,
    "label": labels})
```

```python
review_df.to_csv("reviews.csv", index=False)
print(review_df.head())
print("\nTotal Reviews:", len(review_df))
print("\nSentiment Distribution:")
print(review_df["label"].value_counts())
```
