```python
import pandas as pd
users = pd.read_csv("users.csv")
orders = pd.read_csv("orders.csv")
orders["order_date"] = pd.to_datetime(orders["order_date"])
users["signup_date"] = pd.to_datetime(users["signup_date"])
latest_date = orders["order_date"].max()
customer_features = (orders.groupby("user_id").agg(spend=("price", "sum"),frequency=("order_id", "count"),last_order=("order_date", "max")).reset_index())
customer_features["recency"] = (latest_date - customer_features["last_order"]).dt.days
customer_features["avg_order_value"] = (customer_features["spend"] /customer_features["frequency"])
customer_features = customer_features[
    ["user_id","spend","frequency","recency","avg_order_value"]]
users = users.merge(customer_features, on="user_id", how="left")
users["spend"] = users["spend"].fillna(0)
users["frequency"] = users["frequency"].fillna(0)
users["recency"] = users["recency"].fillna(
    (latest_date - users["signup_date"]).dt.days)
users["avg_order_value"] = users["avg_order_value"].fillna(0)
users.to_csv("uuser.csv", index=False)
print(users.head())
```
